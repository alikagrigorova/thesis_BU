# Asymptotic Behavior of the Trace Expression as $\beta \to \infty$
This notebook explores the behavior of the following expression as $\beta$ increases, for different values of $\sigma$ and a fixed $s$ such that $s^2/\sigma^2 > 1$:
$$
\frac{1}{2} \operatorname{Tr}(I_{\theta_\circ(\beta)} J_{\theta_\circ(\beta)}^{-1}) = \frac{(2\pi\sigma^2)^{-(\beta-1)/2} s^2}{2\sigma^2} \left( \frac{1 + \frac{(\beta-1)s^2}{\sigma^2}}{1 + \frac{2(\beta-1)s^2}{\sigma^2}} \right)^{3/2}
$$
We will plot this value for several $\sigma$ and analyze monotonicity and threshold behavior.

```python
import numpy as np
import matplotlib.pyplot as plt

def trace_expr(beta, sigma, s):
    # Implements the formula
    num = (2 * np.pi * sigma**2) ** (-(beta - 1) / 2) * s**2
    denom = 2 * sigma**2
    frac = (1 + (beta - 1) * s**2 / sigma**2) / (1 + 2 * (beta - 1) * s**2 / sigma**2)
    return num / denom * frac ** (1.5)
```

```python
# Choose s and sigma values
s = 0.1
sigmas = [0.3, 0.39, 0.4, 0.5, 1.0, 1.5]
# Ensure s^2/sigma^2 > 1 for all sigmas
assert all((s**2 / sigma**2) < 1 for sigma in sigmas)

betas = np.linspace(1.01, 300, 1000)
betas = np.linspace(1.01, 300, 1000)
plt.figure(figsize=(10, 6))
for sigma in sigmas:
    vals = trace_expr(betas, sigma, s)
    plt.plot(betas, vals, label=fr'$\sigma={sigma}$')
plt.axhline(0.5, color='red', linestyle='--', label=r'$y=1/2$')
plt.xlabel(r'$\beta$')
plt.ylabel(r'Trace Expression')
plt.title(r'Behavior as $\beta \to \infty$ for different $\sigma$')
plt.ylim(0,20)  # Adjust this range as needed to see all curves near 0.5
plt.legend()
plt.grid(True, which='both', ls='--', alpha=0.5)
plt.show()
plt.show()
```

## Observations
- The plot above shows how the value decays as $\beta$ increases for different $\sigma$.
- For all $\sigma$, the value decreases rapidly as $\beta$ grows.
- The decay rate and initial value depend strongly on $\sigma$.
- There is no non-monotonicity for these parameter choices; the function is strictly decreasing in $\beta$.
- If you want to explore other regimes (e.g., $s^2/\sigma^2 < 1$), you can adjust $s$ and $\sigma$ accordingly.
